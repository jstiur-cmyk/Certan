```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Penjelasan :
Kode ini memanggil library yang dibutuhkan. NumPy dipakai untuk mengolah angka dan Pandas untuk mengolah data berbentuk tabel. Setelah itu, dataset Iris diambil dari Scikit-Learn dengan load_iris() dan disimpan ke variabel iris.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Penjelasan :
Data Iris diubah menjadi tabel (DataFrame), lalu ditambahkan kolom Species berisi nama spesies bunga. X berisi empat ukuran bunga sebagai fitur, sedangkan y berisi nama spesies sebagai target yang akan ditebak. print(X) menampilkan tabel fitur berukuran 150 baris dan 4 kolom.

```python
y
```

Penjelasan :
Kode ini menampilkan isi y, yaitu label spesies setiap bunga. Labelnya masih berupa teks (setosa, versicolor, dan virginica), masing-masing sebanyak 50 data.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Penjelasan :
LabelEncoder mengubah label teks menjadi angka supaya bisa diproses model: setosa menjadi 0, versicolor menjadi 1, dan virginica menjadi 2.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Penjelasan :
Data dibagi menjadi 80% data latih (120 data) dan 20% data uji (30 data). random_state=0 membuat pembagiannya selalu sama, sehingga setiap fitur tetap berpasangan dengan labelnya.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Penjelasan :
Model yang dipakai adalah Gaussian Naïve Bayes, karena semua fitur Iris berupa angka kontinu (bisa bernilai desimal). Saat dilatih dengan fit(), model menghitung rata-rata dan standar deviasi setiap fitur untuk setiap spesies. Nilai-nilai ini nanti dipakai untuk menghitung peluang sebuah bunga termasuk spesies tertentu.

```python
result = NB_model.predict(X_test)
result
```

Penjelasan :
Model menebak spesies dari 30 data uji. Hasilnya berupa angka 0, 1, dan 2 yang mewakili setosa, versicolor, dan virginica.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Penjelasan :
Akurasi dihitung dengan membandingkan hasil tebakan dan label asli, lalu dikali 100 supaya menjadi persen. Hasilnya 96.67%, artinya 29 dari 30 data uji ditebak dengan benar.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Penjelasan :
Kode ini menebak spesies untuk satu bunga baru, lalu inverse_transform() mengubah hasil angkanya kembali menjadi nama spesies. Hasilnya setosa. Di output akan muncul tulisan merah UserWarning. Ini bukan error, hanya peringatan. Penyebabnya, model dilatih memakai DataFrame yang punya nama kolom, sedangkan input di sini berupa list biasa tanpa nama kolom.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Penjelasan :
Ini adalah cara yang benar. Input dibungkus menjadi DataFrame dengan nama kolom yang sama seperti data latih, sehingga warning tidak muncul lagi. Hasilnya tetap setosa.

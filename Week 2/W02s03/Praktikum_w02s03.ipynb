Outliers

```python
path = "weight-height_outlier.csv"
df2 = pd.read_csv(path)

df2
```

```python
import seaborn as sns

sns.boxplot(df2['Height'])
```

```python
import numpy as np

# Identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Height'], [25, 75])

# Hitung IQR
iqr = q3 - q1

# Cari lower dan upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

# Hapus outliers
clean_data = df2[(df2['Height'] >= lower_bound)
                 & (df2['Height'] <= upper_bound)]

sns.boxplot(clean_data['Height'])
```

Mendeteksi outliers dengan visualisasi dari matplotlib.

```python
import matplotlib.pyplot as plt

# Membuat daftar kolom yang ingin dianalisis, yaitu "Weight" dan "Height".
columns_to_plot = ['Weight', 'Height']

# Memastikan kolom ada di dataset
columns_to_plot = [col for col in columns_to_plot if col in df2.columns]

fig, axs = plt.subplots(
    len(columns_to_plot),
    1,
    dpi=95,
    figsize=(7, 5 * len(columns_to_plot))
)

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(df2[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
def remove_outliers_sequentially(df2, columns):
    df_cleaned = df2.copy()

    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1

        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)

        df_cleaned = df_cleaned[
            (df_cleaned[col] >= lower_bound) &
            (df_cleaned[col] <= upper_bound)
        ]

    return df_cleaned
```

```python
# Terapkan fungsi untuk menghapus outliers dari "Height" dan "Weight"
clean_data = remove_outliers_sequentially(df2, ['Height', 'Weight'])

# Plot boxplots setelah menghapus outliers
columns_to_plot = ['Weight', 'Height']

fig, axs = plt.subplots(
    len(columns_to_plot),
    1,
    dpi=95,
    figsize=(7, 5 * len(columns_to_plot))
)

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(clean_data[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0, 1))

# Memisahkan kolom numerik dan non-numerik
numeric_cols = clean_data.select_dtypes(include=[np.number]).columns
non_numeric_cols = clean_data.select_dtypes(exclude=[np.number]).columns

rescaled_numeric = scaler.fit_transform(clean_data[numeric_cols])

rescaled_numeric[:]
```

```python
# Konversi kembali hasil scaling ke DataFrame
rescaled_df = pd.DataFrame(
    rescaled_numeric,
    columns=numeric_cols,
    index=clean_data.index
)

# Gabungkan kembali dengan kolom non-numerik tanpa mengubah jumlah baris
final_df = pd.concat(
    [clean_data[non_numeric_cols], rescaled_df],
    axis=1
)

final_df
```

3. Handling outliers dari dataset 'diabetes.csv'
Pilih 3 kolom untuk deteksi
Bebas menggunakan matplotlib atau seaborn

Step 1 — Load dataset

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df2 = pd.read_csv("diabetes.csv")
df2.head()
```

Step 2 — Deteksi outlier secara visual dengan boxplot (sebelum dibersihkan)

```python
columns_to_plot = ['Insulin', 'BMI', 'BloodPressure']

fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))
for i, col in enumerate(columns_to_plot):
    sns.boxplot(x=df2[col], ax=axs[i])
    axs[i].set_ylabel(col)
plt.show()
```

Step 3 — Hitung IQR dan hapus outlier untuk ketiga kolom sekaligus

```python
def remove_outliers_sequentially(df, columns):
    df_cleaned = df.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)]
    return df_cleaned

clean_data = remove_outliers_sequentially(df2, ['Insulin', 'BMI', 'BloodPressure'])
```

Step 4 — Plot ulang boxplot setelah outlier dihapus, sebagai pembanding

```python
fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))
for i, col in enumerate(columns_to_plot):
    sns.boxplot(x=clean_data[col], ax=axs[i])
    axs[i].set_ylabel(col)
plt.show()
```

Step 5 — Bandingkan jumlah baris sebelum dan sesudah

```python
print("Jumlah baris sebelum:", df2.shape[0])
print("Jumlah baris sesudah:", clean_data.shape[0])
```

Hasil Analisis :
Dari 768 baris data diabetes.csv, setelah dihapus outlier pada kolom Insulin, BMI, dan BloodPressure menggunakan metode IQR, tersisa 677 baris (91 baris terhapus, sekitar 12%). Outlier terbanyak berasal dari nilai 0 pada Insulin dan BloodPressure, yang kemungkinan sebenarnya data hilang, bukan outlier murni. Setelah dibersihkan, data jadi lebih representatif dan siap dipakai untuk analisis lanjutan.
